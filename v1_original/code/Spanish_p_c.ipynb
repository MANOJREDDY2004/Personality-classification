```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
```

The PAN Personality Dataset is a multilingual personality prediction dataset released as part of the PAN Author Profiling tasks. It contains social media texts such as tweets and online posts collected from users in different languages including English and Spanish. Each text sample is labeled with personality traits or personality classes, making it suitable for Natural Language Processing (NLP) and machine learning tasks. The dataset is widely used for author profiling, personality detection, and behavioral text analysis research.

The dataset is widely used for tasks such as:

Personality prediction
Author profiling
Sentiment and behavioral analysis
Text classification
Multilingual NLP research

```python
# Load dataset
df = pd.read_csv("spanish_dataset.csv")
```

```python
# Show first 5 rows
print(df.head())
```

```python
# Input and output
X = df["text"]
y = df["label"]
```

```python
from nltk.corpus import stopwords
import nltk

nltk.download('stopwords')

spanish_stopwords = stopwords.words('spanish')

vectorizer = TfidfVectorizer(stop_words=spanish_stopwords)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Create TF-IDF vectorizer
vectorizer = TfidfVectorizer()

# Convert text into numerical vectors
X_vectorized = vectorizer.fit_transform(X)
```

```python
print(len(vectorizer.vocabulary_))
```

```python
# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X_vectorized,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns

# Model
knn_model = KNeighborsClassifier(n_neighbors=5)

# Train
knn_model.fit(X_train, y_train)

# Predict
knn_pred = knn_model.predict(X_test)

# Accuracy
knn_acc = accuracy_score(y_test, knn_pred)

print("KNN Accuracy:", knn_acc)

# Classification Report
print(classification_report(y_test, knn_pred))

# Confusion Matrix
cm_knn = confusion_matrix(y_test, knn_pred)

plt.figure(figsize=(6,4))
sns.heatmap(cm_knn, annot=True, fmt='d', cmap='Blues')
plt.title("KNN Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
k_values = range(1, 21)
train_acc = []
test_acc = []

for k in k_values:
    model = KNeighborsClassifier(n_neighbors=k)

    model.fit(X_train, y_train)

    train_acc.append(model.score(X_train, y_train))
    test_acc.append(model.score(X_test, y_test))

plt.figure(figsize=(8,5))

plt.plot(k_values, train_acc, label='Train Accuracy')
plt.plot(k_values, test_acc, label='Test Accuracy')

plt.xlabel("K Value")
plt.ylabel("Accuracy")
plt.title("KNN Accuracy for Different K Values")

plt.legend()
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression

# Model
lr_model = LogisticRegression(max_iter=1000)

# Train
lr_model.fit(X_train, y_train)

# Predict
lr_pred = lr_model.predict(X_test)

# Accuracy
lr_acc = accuracy_score(y_test, lr_pred)

print("Logistic Regression Accuracy:", lr_acc)

# Classification Report
print(classification_report(y_test, lr_pred))

# Confusion Matrix
cm_lr = confusion_matrix(y_test, lr_pred)

plt.figure(figsize=(6,4))
sns.heatmap(cm_lr, annot=True, fmt='d', cmap='Blues')
plt.title("Logistic Regression Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
c_values = [0.01, 0.1, 1, 10, 100]
lr_scores = []

for c in c_values:

    model = LogisticRegression(
        C=c,
        max_iter=1000
    )

    model.fit(X_train, y_train)

    score = model.score(X_test, y_test)

    lr_scores.append(score)

plt.figure(figsize=(8,5))

plt.plot(c_values, lr_scores, marker='o')

plt.xscale('log')

plt.xlabel("C Value")
plt.ylabel("Accuracy")

plt.title("Logistic Regression Accuracy")

plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier

# Model
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train
rf_model.fit(X_train, y_train)

# Predict
rf_pred = rf_model.predict(X_test)

# Accuracy
rf_acc = accuracy_score(y_test, rf_pred)

print("Random Forest Accuracy:", rf_acc)

# Classification Report
print(classification_report(y_test, rf_pred))

# Confusion Matrix
cm_rf = confusion_matrix(y_test, rf_pred)

plt.figure(figsize=(6,4))
sns.heatmap(cm_rf, annot=True, fmt='d', cmap='Blues')
plt.title("Random Forest Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
tree_values = [10, 50, 100, 150, 200]
rf_scores = []

for trees in tree_values:

    model = RandomForestClassifier(
        n_estimators=trees,
        random_state=42
    )

    model.fit(X_train, y_train)

    score = model.score(X_test, y_test)

    rf_scores.append(score)

plt.figure(figsize=(8,5))

plt.plot(tree_values, rf_scores, marker='o')

plt.xlabel("Number of Trees")
plt.ylabel("Accuracy")

plt.title("Random Forest Accuracy")

plt.show()
```

```python
models = [
    'KNN',
    'Logistic Regression',
    'Random Forest'
]

accuracies = [
    knn_acc,
    lr_acc,
    rf_acc
]

plt.figure(figsize=(8,5))

bars = plt.bar(models, accuracies)

plt.xlabel("Algorithms")
plt.ylabel("Accuracy")

plt.title("Model Accuracy Comparison")

for bar in bars:

    yval = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width()/2,
        yval + 0.01,
        round(yval, 3),
        ha='center'
    )

plt.show()
```
